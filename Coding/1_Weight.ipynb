```python
#extensiones necesarias:
import numpy as np 
from numpy import array, dot, zeros, matmul, ndarray   
import scipy as sp 
from scipy.linalg import lu_factor, lu_solve, solve, norm 
import itertools as it 
#definimos también la matriz A y los vectores x y b
B=array([[1, 1, 1],
         [1, 1, -1],
         [1, -2, 0]])
v=array([50000, 15000, 0])

#Método de Gauss:
def pivoting_row_swapping(A,b,k,N):
    s=zeros(N)
    for i in range(k, N):   # loop over rows
       s[i] = max( abs(A[i,k:N]) ) 
    pivot=abs( A[k,k] / s[k] ) #    find a row with the largest pivoting element
    l = k
    for j in range(k, N): 
       if abs( A[j,k] / s[j] ) > pivot: 
          pivot = abs( A[j,k] / s[j] )
          l = j
#    Check if the system has a sigular matrix
    if pivot == 0.0 :  print(" The matrix is singular "); exit()
#    pivoting: swap rows k and l (if needed)
    if l != k : 
        A[ [k, l] ,k:N] =  A[ [l, k], k:N]  
        ( b[k], b[l] ) =  ( b[l], b[k] )

        
def my_Gauss(A:ndarray,b:ndarray)->ndarray:
    N=len(b); x=zeros(N)
    for i in range(0,N):
        pivoting_row_swapping(A,b,i,N)
        for k in range(i+1,N):
            c=A[k,i]/A[i,i]
            A[k,:]=A[k,:]-c*A[i,:]
            b[k]=b[k]-c*b[i]
    for i in range(N-1,-1,-1):
        x[i]=(b[i]-dot(A[i,i+1:N],x[i+1:N]))/A[i,i]
    return x
print(my_Gauss(B,v))

solve(B,v)
```

```python
#método de Gauss_Seidel
def my_GaussSeidel(A,b):
  N = len(b)
  x1, x2, b1 = zeros(N), zeros(N), zeros(N)
  Error, tolerance = 1, 1e-10 
  k, max_iter = 0, 100

  while Error > tolerance  and  k < max_iter: 
    x1[:] = x2[:]   # x1=x2 does not work 
    for i in range(N):
      b1[i] = b[i] - dot( A[i, i+1:], x1[i+1:] )  # old values  
    for i in range(N):
      x2[i]  = ( b1[i] - dot( A[i, :i], x2[:i] ) )/ A[i, i] 
         
    Error = norm(x2-x1)
    k += 1

  return x2     
print(my_GaussSeidel(B,v))
```

```python
#Método de Jacobi:
def my_Jacobi(A,b):
    N = len(b)
    x, D, R = zeros(N), zeros(N), zeros((N,N))
    xn = 1 + x 
    tolerance = 1e-10
    k, max_iter = 0, 100

    for i in range(N): 
        for  j in range(N): 
            if i==j: 
                D[i] = A[i,j]
                R[i,j] = 0 
            else:
                R[i,j] = A[i,j]
   
    while norm(xn - x) > tolerance  and  k < max_iter: 
        x = xn
        xn = (b - dot(R, x)) / D
      

    return x 

print(my_Jacobi(B,v))
```
