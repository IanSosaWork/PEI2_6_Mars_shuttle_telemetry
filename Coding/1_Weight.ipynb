```python
#Resolución del sistema de ecuaciones que relaciona las masas m1, m2, m3 de los tanques de combustible
#de la aeronave, se plantea como un sistema matricial Ax=b con las relaciones m1+m2+m3=50000, m1+m2-m3=15000
#y m1-2m2=0
#tendremos disponibles una serie de métodos de resolución:
```

```python
#extensiones necesarias:
import numpy as np 
from numpy import array, dot, zeros, matmul, ndarray    
import scipy as sp 
from scipy.linalg import lu_factor, lu_solve, solve
import itertools as it 
#definimos también la matriz A y los vectores x y b
B=array([[1, 1, 1],
         [1, 1, -1],
         [1, -2, 0]])
v=array([50000, 15000, 0])

#Método de Gauss:
def my_Gauss(A:ndarray,b:ndarray)->ndarray:
    N=len(b); x=zeros(N)
    for i in range(0,N):
        for k in range(i+1,N):
            c=A[k,1]/A[i,i]
            A[k,:]=A[k,:]-c*A[i,:]
            b[k]=b[k]-c*b[i]
    for i in range(N-1,-1,-1):
        x[i]=(b[i]-dot(A[i,i+1:],x[i+1:]))/A[i,i]
    return x
print(my_Gauss(B,v))
```
